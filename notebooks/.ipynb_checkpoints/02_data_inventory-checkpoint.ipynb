```python
from pathlib import Path

# Find the raw-data folder from the current directory
candidates = [
    Path.cwd() / "data" / "raw",
    Path.cwd().parent / "data" / "raw",
]

raw_dir = next(
    (path for path in candidates if path.is_dir()),
    None
)

if raw_dir is None:
    raise FileNotFoundError(
        "Could not find data/raw. Check your project folder."
    )

excel_files = sorted(
    list(raw_dir.glob("*.xlsx"))
    + list(raw_dir.glob("*.xls"))
)

print(f"Raw data folder: {raw_dir}")
print(f"Excel files found: {len(excel_files)}")

for file in excel_files:
    print(f"{file.name} — {file.stat().st_size / 1_000_000:.2f} MB")
```

```python
import pandas as pd
from pathlib import Path

# Define the study period
START_YEAR = 2011
END_YEAR = 2026

# Find the three source workbooks
excel_files = sorted(raw_dir.glob("*.xlsx"))

# Store summary results here
inventory = []

for file_path in excel_files:
    with pd.ExcelFile(file_path) as xls:

        for sheet in xls.sheet_names:

            # Ignore reference sheets such as "Drug list and Measurement Units"
            if not sheet.isdigit():
                continue

            year = int(sheet)

            # Limit the study period to 2011–2026
            if not START_YEAR <= year <= END_YEAR:
                continue

            print(f"Inspecting {year}...")

            # Load one annual sheet at a time
            data = pd.read_excel(
                xls,
                sheet_name=sheet
            )

            # Standardize column labels for comparison
            data.columns = (
                data.columns
                .str.strip()
                .str.replace("Subrgion", "Subregion", regex=False)
            )

            # Convert dates consistently for profiling
            dates = pd.to_datetime(
                data["Seizure Date"],
                errors="coerce"
            )

            # Collect a summary of this year's data
            inventory.append({
                "year": year,
                "file": file_path.name,
                "rows": len(data),
                "columns": len(data.columns),
                "date_min": dates.min(),
                "date_max": dates.max(),
                "unique_dates": dates.nunique(),
                "missing_dates": dates.isna().sum(),
                "missing_countries": (
                    data["ISO3"].isna().sum()
                ),
                "missing_drugs": (
                    data["Drug/Substance"].isna().sum()
                ),
                "missing_quantities": (
                    data["Quantity Seized"].isna().sum()
                ),
                "nonpositive_quantities": (
                    data["Quantity Seized"].le(0).sum()
                ),
                "exact_duplicate_rows": (
                    data.duplicated().sum()
                ),
                "unique_drugs": (
                    data["Drug/Substance"].nunique()
                ),
                "unique_countries": (
                    data["ISO3"].nunique()
                )
            })

            del data

# Create the inventory table
inventory_df = (
    pd.DataFrame(inventory)
    .sort_values("year")
    .reset_index(drop=True)
)

display(inventory_df)
```

```python
# Load one year with a high duplicate count
with pd.ExcelFile(
    raw_dir / "IDS-data-2011_17-May24.xlsx"
) as xls:

    df_2014 = pd.read_excel(
        xls,
        sheet_name="2014"
    )

# Count exact duplicate rows
duplicate_mask = df_2014.duplicated(keep=False)

print("Total records:", len(df_2014))
print("Rows belonging to duplicate groups:", duplicate_mask.sum())
print("Excess duplicate rows:", df_2014.duplicated().sum())

# Show examples of repeated records
display(
    df_2014.loc[duplicate_mask]
    .sort_values(df_2014.columns.tolist())
    .head(20)
)
```

```python
import pandas as pd

START_YEAR = 2011
END_YEAR = 2026

date_checks = []

for file_path in sorted(raw_dir.glob("*.xlsx")):

    with pd.ExcelFile(file_path) as xls:

        for sheet in xls.sheet_names:

            if not sheet.isdigit():
                continue

            year = int(sheet)

            if not START_YEAR <= year <= END_YEAR:
                continue

            data = pd.read_excel(
                xls,
                sheet_name=sheet,
                usecols=["Seizure Date"]
            )

            dates = pd.to_datetime(
                data["Seizure Date"],
                errors="coerce"
            ).dropna()

            if dates.empty:
                continue

            date_counts = dates.value_counts()
            most_common_date = date_counts.index[0]

            date_checks.append({
                "year": year,
                "total_records": len(data),
                "valid_dates": len(dates),
                "missing_or_invalid_dates": data["Seizure Date"].isna().sum()
                    + pd.to_datetime(
                        data["Seizure Date"],
                        errors="coerce"
                    ).isna().sum()
                    - data["Seizure Date"].isna().sum(),
                "dec_31_count": (
                    (dates.dt.month == 12) &
                    (dates.dt.day == 31)
                ).sum(),
                "dec_31_pct": round(
                    (
                        (dates.dt.month == 12) &
                        (dates.dt.day == 31)
                    ).mean() * 100,
                    3
                ),
                "most_common_date": most_common_date.date(),
                "most_common_date_count": int(date_counts.iloc[0]),
                "most_common_date_pct": round(
                    date_counts.iloc[0] / len(dates) * 100,
                    3
                )
            })

date_checks_df = (
    pd.DataFrame(date_checks)
    .sort_values("year")
    .reset_index(drop=True)
)

display(date_checks_df)
```

```python
from pathlib import Path

processed_dir = raw_dir.parent / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

print("Processed data folder:", processed_dir)
```

```python
import pandas as pd

START_YEAR = 2011
END_YEAR = 2026

frames = []

for file_path in sorted(raw_dir.glob("*.xlsx")):

    with pd.ExcelFile(file_path) as xls:

        for sheet in xls.sheet_names:

            if not sheet.isdigit():
                continue

            year = int(sheet)

            if not START_YEAR <= year <= END_YEAR:
                continue

            print(f"Loading {year}...")

            data = pd.read_excel(
                xls,
                sheet_name=sheet
            )

            # Standardize column names
            data.columns = data.columns.str.strip()

            data = data.rename(columns={
                "Subrgion": "Subregion"
            })

            # Preserve source information
            data["source_year"] = year
            data["source_file"] = file_path.name
            data["source_sheet"] = sheet

            frames.append(data)

# Combine all annual sheets
ids = pd.concat(
    frames,
    ignore_index=True,
    sort=False
)

print("Combined rows:", len(ids))
print("Combined columns:", len(ids.columns))
print("Years:", ids["source_year"].min(), "-", ids["source_year"].max())

display(ids.head())
```

```python
# Parse dates and quantities
ids["Seizure Date"] = pd.to_datetime(
    ids["Seizure Date"],
    errors="coerce"
)

ids["Quantity Seized"] = pd.to_numeric(
    ids["Quantity Seized"],
    errors="coerce"
)

# Remove leading/trailing spaces from text fields
text_cols = ids.select_dtypes(
    include=["object", "string"]
).columns

for col in text_cols:
    ids[col] = ids[col].str.strip()

# Save the consolidated dataset
output_path = processed_dir / "ids_2011_2024.parquet"

ids.to_parquet(
    output_path,
    index=False
)

print("Saved to:", output_path)
print("Rows:", len(ids))
print("Columns:", len(ids.columns))
```
